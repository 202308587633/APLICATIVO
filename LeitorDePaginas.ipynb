```python
# Importações
import os
import re
from bs4 import BeautifulSoup
from dotenv import load_dotenv
from selenium import webdriver
from selenium.webdriver.edge.options import Options
```

```python
# Lê variáveis de ambiente do arquivo .env
load_dotenv()
```

```python
# Função para ler página do agregador de trabalhos acadêmicos
# url_agregador = input("Digite a URL do agregador de trabalhos acadêmicos: ")
url_agregador = "https://bdtd.ibict.br/vufind/Search/Results?join=AND&bool0%5B%5D=AND&lookfor0%5B%5D=%22an%C3%A1lise+de+discurso%22&type0%5B%5D=AllFields&lookfor0%5B%5D=direito&type0%5B%5D=AllFields&illustration=-1&daterange%5B%5D=publishDate&publishDatefrom=2021&publishDateto=2021"
# url_agregador = "https://bdtd.ibict.br"

edge_options = Options()
edge_options.add_argument("--headless=new") # Esta linha faz a mágica de esconder a janela
edge_options.add_argument("--disable-gpu")  # Recomendado para evitar erros em modo headless
edge_options.add_argument("--window-size=1920,1080") # Garante que o site carregue o layout de desktop

try:
    driver = webdriver.Edge(options=edge_options)
    driver.get(url_agregador)
    driver.implicitly_wait(10)
    response_text = driver.page_source
    driver.quit()
except Exception as e:
    print(f"Erro: {e}")
```

```python
# Trata a página lida para recuperar os dados dos trabalhos
soup = BeautifulSoup(response_text, 'html.parser')
```

```python
# Função para extrair dados dos trabalhos acadêmicos
def extrair_dados_trabalhos(soup):
    trabalhos_detalhados = []

    # No BDTD (VuFind), cada trabalho fica dentro de um bloco com a classe 'result'
    blocos_trabalhos = soup.find_all('div', class_='result-body')    

    for bloco in blocos_trabalhos:
        # Analisando o código HTML da página, identificamos os elementos que contêm as informações desejadas
        titulo = bloco.find_all('a')[0].get_text()
        link = bloco.find('a', class_='fulltext')['href']
        autor = bloco.find_all('a')[1].get_text()

        # Armazenar em um dicionário
        info_trabalho = {
            "Título": titulo,
            "Autor": autor,
            "Link": link
        }
        trabalhos_detalhados.append(info_trabalho)
```
